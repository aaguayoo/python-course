## Declaración de clases

Para poder declarar una clase en Python, se cuenta con la partícula __class__
```python
class NombreClase:
    ···
```

La estructura de clase más básica es la siguiente:

````python
# Objeto bicicleta
class Bicicleta:
    pass
````

Esto nos permitiría crear objetos a partir de la clase `Bicicleta`, sin embargo serán objetos sin ninguna capacidad (métodos) o características (atributos). Para crear un objeto, es necesario _inicializar_ la clase, y para hacerlo necesitamos declarar una variable desde nuestra clase:

````python
# objeto = Clase()
bicicleta = Bicicleta()
````

Declaremos ahora un par de _atributos de instancia_ a nuestra clase, es decir, atributos cuyo valor puede variar entre diferentes objetos, para hacer esto podemos hacer uso del método mágico `.__init__()`

````python
class Bicicleta:
    
    def __init__(self, color:str , rodada: int):
        self.color = color
        self.rodada = rodada
````

Notemos que hemos declarado un argumento `self` como el primer argumento de la función `__init__`, esto es necesario ya que será a través de este objeto que compartiremos información entre todos los métodos internos de la clase, es decir, que todo aquello que sea declarado en el objeto self., podrá tener alcances globales dentro de la clase.

Ahora al instanciar nuestro objeto, debemos de pasarle todos los argumentos que pida el método `__init__`para poder inicializar.

````python
bicicleta_montaña = Bicicleta(color="naranja", rodada=29)
bicicleta_ruta = Bicicleta(color="gris", rodada=27)
````

````python
bicicleta_montaña.color, bicicleta_ruta.color
````

Esto hará que nuestro objeto `bicicleta_montaña` posea atributos `.color` y `.rodada`. 

Podemos ahora declarar _atributos de clase_, los cuales son atributos que entre objetos no varían su valor

````python
print(bicicleta_montaña)
````

````python
class Bicicleta:
    
    numero_ruedas: int = 2
    
    def __init__(self, color:str , rodada: int):
        self.color = color
        self.rodada = rodada
````

````python
bicicleta_montaña = Bicicleta(color="naranja", rodada=29)
bicicleta_ruta = Bicicleta(color="gris", rodada=27)
````

````python
bicicleta_montaña.numero_ruedas
````

````python
bicicleta_ruta.numero_ruedas
````

## Documentación de clase

Cuando hemos realizado una impresión en pantalla de nuestro objeto `bicicleta`hemos visto que el restulado no es humanamente entendible, por lo que podemos agregar un método mágico `.__str__()` a nuestra clase para crear mejores descripciones.

````python
class Bicicleta:
    
    numero_ruedas: int = 2
    
    def __init__(self, color:str , rodada: int):
        self.color = color
        self.rodada = rodada
        
    def __str__(self):
        return f"Bicicleta de rodada {self.rodada} y color {self.color}"
````

````python
bicicleta_montaña = Bicicleta(color="naranja", rodada=29)
bicicleta_ruta = Bicicleta(color="gris", rodada=27)
````

````python
print(bicicleta_montaña)
````

````python
print(bicicleta_ruta)
````

## Clases como agrupación de funciones

En los casos anteriores el uso de la clase era especificamente para el cumplimiento del paradigma de programación en Python, que es la POO (Programación Orientada a Objetos) en el cual las clases nos permiten crear objetos que corresponden a tipos de datos más completos y personalizables, usando como base los que ya tiene pre-definido Python.

Pero por otro lado, una clase nos puede funcionar para agrupar un gran conjunto de funcionalidades complejas que quieran ser aplicado a un mismo grupo de argumentos de entrada, como se muestra a continuación:

````python
def suma(*args):
    return sum(args)
````

````python
def multiplicacion(*args):
    total = 1
    for arg in args:
        total *= arg
    return total
````

````python
class Operaciones:
    
    def __init__(self, *args):
        self.args = args
        
    def suma(self):
        return sum(self.args)
    
    def multiplicacion(self):
        total = 1 
        for arg in self.args:
            total *= arg
        
        return total
````

````python
operacion = Operaciones(1,2,3,4)
````

````python
operacion.suma()
````

````python
operacion.multiplicacion()
````
