# Métodos en Clases

Hemos visto que podemos crear estructuras de datos a través de las clases, sin embargo, el delcarar clasese tiene alcances mayores, uno de esos alcances es determinar capacidades del objeto a través de los _métodos_.

Los métodos son _funciones_ internas de la clase, las cuales se comportan exactamente igual a una función convencional (cumplen las mismas reglas de argumentos pre-definidos y posicionales, y `*args`, y `*kwargs`, y tipados). La única diferencia es que su primer argumento (casi siempre) es la partícula *self*.

```python
class Bicicleta:
	número_de_ruedas = 2

    # Método oculto (privado) pre-definido de Python, el más importante para inicializar un objeto
	def __init__(self, color, tipo):
		self.color = color
		self.tipo = tipo

    # Método oculto pre-definido de Python
	def __str__(self):
		return f"Bicicleta de {self.tipo}, color {self.color}"
```

Para crear métodos propios:
```python
class NombreClase:
	
	...

	def _método_privado_1(self, arg_1_1, arg_2_1):
		sentencias

	def método_1(self, arg_1_1, arg_2_1):
		sentencias

	def método_2(self, arg_1_2, arg_2_2, arg_3_2):
		sentencias
```

Si bien los métodos se pueden pensar como funciones de una clase, es mejor pensarlos como capacidades que puede realizar un objeto en cuestión. Una manera burda de saber si es una capacidad o no, para saber si se implementa como método o no, es pensar si dicha capacidad se define como una acción o un verbo.

````python
# Ejemplo
class Carro:
    
    # Atributos de clase
    numero_ruedas = 4
    tieneMotor = True
    numero_puertas = 4
    estaEncendido = False
    
    # Atributos de instancia
    def __init__(self, color, caballos_de_fuerza):
        self.color = color
        self.caballos_de_fuerza = caballos_de_fuerza
        
    # El método recibe los atributos (clase, instancia) y los modifica
    def encender_carro(self):
        self.estaEncendido = True
        
    def avanzar(self, velocidad):
        if self.estaEncendido:
            smog_generado = self.caballos_de_fuerza * velocidad
            return smog_generado
        else:
            print("El carro está apagado")
    
    def _llenar_aceite(self):
        print("Cantidad de aceite")
````

````python
carrito = Carro(
    color="negro",
    caballos_de_fuerza=200,
)
````

````python
carrito.encender_carro()
````

````python
carrito.avanzar(120)
````

````python
carrito.
````

````python
otro_carrito = Carro(
    color="rojo",
    caballos_de_fuerza=100,
)
````

````python
otro_carrito.encender_carro()
````

````python
otro_carrito.avanzar(120)
````

## Tipos de métodos

Existen 4 tipos de métodos que se pueden definir en una clase. El primero, y que ya vimos son los métodos convencionales (ya sea públicos o privados), los cuales como primer argumento siempre requieren la partícula `self`.

## Método convencional
```python
class Class:
    ··· 
    
    def metodo_convencional(self, *args, **kwargs):
        sentencia
```

## Classmethod
```python
class Class:
    ··· 
    
    @classmethod
    def class_method(cls, *args, **kwargs):
        sentencia
``` 

## Staticmethod
```python
class Class:
    ··· 
    
    @staticmethod
    def static_method(*args, **kwargs):
        sentencia
``` 

## Property 
```python
class Class:
    ··· 
    
    @property
    def property(self, *args, **kwargs):
        sentencia
``` 

## Classmethod

Es un método que permite realizar una inicialización de clase alternativa. De esta forma se pueden crear objetos, del mismo tipo, pero con algún tipo de diferencia al original.

````python
class Carro:
    
    estaEncendido = False
    lucesEncendidas = False
    
    def __init__(self):
        # El carro se enciende en la inicialización
        self.estaEncendido = True
        self.lucesEncendidas = True
        
    @classmethod
    def encendido_parcial(cls):
        cls.estaEncendido = False
        cls.lucesEncendidas = True
        return cls
        
````

````python
carro_encendido_total = Carro()
print(carro_encendido_total.estaEncendido)
````

````python
carro_encendido_parcial = Carro().encendido_parcial()
print(carro_encendido_parcial.estaEncendido)
````

## Staticmethod

Este consiste en un método simple que no requiere la partícula `self`, es decir, que no requiere nigún método ni atributo de la clase para realizar una acción.

````python
class Carro:
    
    estaEncendido = False
    lucesEncendidas = False
   
    @staticmethod
    def limpiar_parabrisas():
        print("Limpiando parabrisas")
````

````python
carrito = Carro()
````

````python
carrito.limpiar_parabrisas()
````

## Property

Este tipo de métodos se llaman como si fueran un atributo, a pesar de ser un método. Es decir, para llamaro no es necesario utilizar los paréntesis

````python
class Carro:
    
    estaEncendido = False
    lucesEncendidas = False
    
    def __init__(self, tipo_rines):
        self._tipo_rines = tipo_rines
        
    @property
    def rines(self):
        return self._tipo_rines
````

````python
carrito = Carro(tipo_rines="aluminio")
````

````python
carrito.rines
````
