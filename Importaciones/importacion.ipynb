# Importaciones en Python

**NOTA**: Para Python se hace uso de las palabras _módulos (module)_ o _paquetes (package)_ de forma indistinta, haciendo referencia a lo mismo, que son programas creados en Python que tienen distintas funcionalidades.

En un punto del desarrollo de software, no muy lejano al inicio, nos veremos en la necesidad de crear cosas cada vez más elaboradas o realizar tareas específicas, por ejemplo, leer una variable de entorno, hacer configuraciones al sistema, declarar clases de forma simplificada, etc. Muchas de estas tareas ya han sido realizadas por alguien en el pasado, y han creado paquetes completos donde podemos encontrar esas funciones.

**_Buena práctica_**: Todas las importaciones necesarias en un archivo debne de ser declaradas en la parte superior del código o notebook.
