# Ciclos indefinidos: _while_

Cuando hablamos de iteraciones, nos referimos a la acción de ejecutar un mismo bloque de código _n_-veces. El conjunto de iteraciones lo llamamos _bucle_, y pueden existir en dos tipos de bucles, por un lado, aquellos que están limitados en el número de iteraciónes que es ejecutaod, los bucles definidos (i.e., _for_) y por otor lado aquelos que se repetirán mietnras alguna condición se cumpla, a estos últimos se les conoce como blues o ciclos indefinidos (i.e., _while_).

## Estructura de un ciclo indefinido

Un ciclo indefinido requiere una condición (al igual que en el if) que estará siendo evaluada a cada paso del ciclo. Si esta condición es verdadera (`True`) entonces se ejectuará nuevamente el bloque de código contenido dentro de la estructura:

```python
while condicion_a_evaluar:
    ···
    ···
    ···
    condicon_a_evaluar = ··· # Siempre debe de haber una sentencia que modifique la condición a evaluar
    ···
    ···
    ···
```

La `candicion_a_evaluar` tiene simepre por resultado un valor booleano.

**NOTA**: Si la condición siempre es verdadera entonces el ciclo nunca terminará, por lo que deberá ser interrumpida la ejecución del programa de forma manual.

````python
# Ejemplo de loop infinito
while True:
    print("Hola")
````

El bloque de código anterior no se detendrá de forma automática, por lo que es necesario detenerlo manualmente, ya sea parando el Kernel (para Jupyter) o usando Ctrl + C si el programa corre en terminal.

Para evitar generar ciclos infinitos es necesario que exista alguna declaración para actualizar la condición a evaluar:

````python
n = 0

while n < 5:
    print(n)
    n += 1
    
print("Se sale del while")
````

## Ciclos indefinidos en colecciones

Otro caso de uso de los ciclos indefinidos es el ejecutar un bloque de código mientras existan elementos en una colección, por ejemplo:

````python
enteros = [1, 2, 3, 4, 5]

while enteros:
    print(enteros)
    enteros.pop(0) # El método .pop() quita elementos por índice
````

````python
# Que pasa si queremos hacerlo con un for
enteros = [1,2,3,4,5]
for _ in enteros:
    print(enteros)
    enteros.pop(0)
````

**NOTA**: Siempre que se utilice una colección como condición a evaluar, esta regresará True si la colección tiene elemtnos y regresará FAlse si la colección es vacía.

````python
# Ejemplo con condicional anidado
enteros = [1,2,3,4]

while enteros:
    entero = enteros.pop(0) # Asignarle a la variable entero el primer elemento de la lista en cada ciclo
    
    if entero % 2 == 0:
        print(f"{entero} es par")
````

## Interrupción de ciclos

El incumplimento de la condición deu n ciclo indefinido no es la única forma en que podemos detener el proceso, también es posible utilizar la partícula **break**, la cual una vez ejecutada, el ciclo se detendra. Esta partícula funciona también para los ciclos definidos.

````python
# Ejemplo con _for_

enteros = [1, 2, 3, 4]

for numero in enteros:
    print(numero)
    
    # Escribir condición de interrupción si el número que se imprimió es par
    if numero % 2 == 0:
        print("El ciclo se va a detener")
        break
````

````python
# Ejemplo con _while_
enteros = [1, 2, 3, 4]

while enteros:
    numero = enteros.pop(0) # Extraemos el primer elemento de la lista y lo quitamos de la misma
    
    print(numero)
    if numero % 2 == 0:
        print("El ciclo se va a detener")
        break
````

## Continuación del ciclo

Por otrlaod, tenemos el caso donde queremos que la iteración sea interrumpida pero sin romper el ciclo, para esto exista la partícula **continue**, la cual permite interrumpir la iteración corriente sin detener todo el ciclo.

````python
# Ejemplo con _for_
enteros = [1, 2, 3, 4]

# No imprimir si el número es par
for numero in enteros:
    
    if numero % 2 == 0:
        continue
        
    print(numero)
````

````python
# Ejemplo con _while_
enteros = [1, 2, 3, 4]

while enteros:
    numero = enteros.pop(0) # Agarramos el primer elemento y lo quitamos de la lista
    
    if numero % 2 == 0:
        continue
        
    print(numero)
````

## Conclusión del ciclo

Es importante poder distinguir los casos entre que un ciclo es concluido debido a una interrupción o si la condición que involucraba el ciclo dejó de ser verdadera, para esto es posible utilzar la partícula **else** junto con el **while**.

Supongamos el caso de un ciclo indefinido donde se suman los valores de una colección hasta agotarla. Queremos adicionalmente detener el ciclo si la suma de los elementos excede un valor dado.

````python
cantidades = [10, 20, 30, 40, 50, 60]
total = 0 # Aquí se va a ir sumando el resultado
limite = 100
````

````python
while cantidades:
    total += cantidades.pop(0) # A total le sumamos el primer elemento de la lista, y lo borramos de la misma
    
    if total > limite:
        print("Sí entro al break")
        break
        
    print(total)
````

Para este ciclo, habrá una interrupción en la 5ta iteración ya que el total ha excedido el límite. Por otro lado, si la colección de cantidades sólo tuviera los primeros cuatro elementos, entonces la salida del ciclo sería idéntica, sin embargo no se detendría el ciclo debido a la partícula **brea**, sino a que ya no se cumpliría la condición del **while**

````python
cantidades = [10, 20, 30, 40]
total = 0
limite = 100

while cantidades:
    total += cantidades.pop(0) # A total le sumamos el primer elemento de la lista, y lo borramos de la misma
    
    if total > limite:
        print("Si entro al break")
        break
        
    print(total)
````

````python
cantidades = [10, 20, 30, 40, 50, 60]
total = 0
limite = 100

while cantidades:
    total += cantidades.pop(0) # A total le sumamos el primer elemento de la lista, y lo borramos de la misma
    
    if total > limite:
        print("Si entro al break")
        break
        
    print(total)
else:
    print("No quedan más elementos en la lista.")
````

Dado que _while_ funciona con una condición, _else_ es importante porque nos otorgaría una respuesta por defecto.

````python
n = 0

while n > 2:
    print("Estoy en el loop")
else:
    print("n no cumple la condición")
````
