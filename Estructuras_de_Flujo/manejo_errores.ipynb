# Manejo de errores

Cuando desarrollamos un código debemos tener en cuenta que habrá casos donde se generen errores, ya sea por una mala codificación o por algún evento inesperado, por ejemplo:

```python
nombre = "Jesús"
edad = 19

print(nombre + edad) # Regresa error porque no se puede sumar un str con un int
```

**NOTA**: Los errores tienen un nombre específico. Este siempre es denotado al final del error, y antes de los ":". Por ejemplo, en el caso anterior se dice que se regresó un TypeError.

Tipos de error predefinidos en Python:
- ValueError -> El valor es incorrecto.
- TypeError -> Inconsistencias con los tipados de variables
- RuntimeError -> Error que tiene que ver con el sistema (CPU o GPU)
- SyntaxError -> Problema de sintaxis de código

https://www.tutorialsteacher.com/python/error-types-in-python

Python al ser un interprete de línea, cuando surjan errores inmediatamente buscará un manejo de error codificado, en caso contrario detendrá el proceso. Esto último no es ideal, ya que quizás se quiere continuar el programa aún suscitdados estos eventos.

Para poder manejar los errores, es decir, prevenir una interrupción abrupta en la ejecución del código, podemos hacer uso de las partículas **try** y **except**.

```python
## Ejemplo usando Exception, donde no importa el tipo de error que salga. Esta es la exception por defecto
num = 10
denom = 0
try:
    resultado = num / denom
    print(resultado)
except Exception:
    print("El código regresó un error")
```

```python
try:
    resultado = num / denom
    print(resultado)
except ZeroDivisionError:
    print("El error que saca es ZeroDivisionError")
    resultado = num / (denom + 0.0001)
    print(resultado)
except ValueError:
    print("El error que saca es ValueError")
    
print("El codigo continua")
print(resultado)
```

**NOTA**: A diferencia de partículas como **if** o **while** donde pueden ser declaradas sin us partículas ligadas como lo son **elif** o **else**, las partículas **try** necesariamente requieren de la existencia de un **except**.
