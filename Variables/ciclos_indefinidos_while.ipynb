# Ciclos indefinidos: _while_

Cuando hablamos de iteraciones, nos referimos a la acción de ejecutar un mismo bloque de código _n_-veces. El conjunto de iteraciones lo llamamos _bucle_, y pueden existir en dos tipos de bucles, por un lado, aquellos que están limitados en el número de iteraciónes que es ejecutaod, los bucles definidos (i.e., _for_) y por otor lado aquelos que se repetirán mietnras alguna condición se cumpla, a estos últimos se les conoce como blues o ciclos indefinidos (i.e., _while_).

## Estructura de un ciclo indefinido

Un ciclo indefinido requiere una condición (al igual que en el if) que estará siendo evaluada a cada paso del ciclo. Si esta condición es verdadera (`True`) entonces se ejectuará nuevamente el bloque de código contenido dentro de la estructura:

```python
while condicion_a_evaluar:
    ···
    ···
    ···
    condicon_a_evaluar = ··· # Siempre debe de haber una sentencia que modifique la condición a evaluar
    ···
    ···
    ···
```

La `candicion_a_evaluar` tiene simepre por resultado un valor booleano.

**NOTA**: Si la condición siempre es verdadera entonces el ciclo nunca terminará, por lo que deberá ser interrumpida la ejecución del programa de forma manual.

````python
# Ejemplo de loop infinito
while True:
    print("Hola")
````

El bloque de código anterior no se detendrá de forma automática, por lo que es necesario detenerlo manualmente, ya sea parando el Kernel (para Jupyter) o usando Ctrl + C si el programa corre en terminal.

Para evitar generar ciclos infinitos es necesario que exista alguna declaración para actualizar la condición a evaluar:

````python
n = 0

while n < 5:
    print(n)
    n += 1
    
print("Se sale del while")
````

## Ciclos indefinidos en colecciones

Otro caso de uso de los ciclos indefinidos es el ejecutar un bloque de código mientras existan elementos en una colección, por ejemplo:

````python
enteros = [1, 2, 3, 4, 5]

while enteros:
    print(enteros)
    enteros.pop(0) # El método .pop() quita elementos por índice
````

````python
# Que pasa si queremos hacerlo con un for
enteros = [1,2,3,4,5]
for _ in enteros:
    print(enteros)
    enteros.pop(0)
````

**NOTA**: Siempre que se utilice una colección como condición a evaluar, esta regresará True si la colección tiene elemtnos y regresará FAlse si la colección es vacía.

````python
# Ejemplo con condicional anidado
enteros = [1,2,3,4]

while enteros:
    entero = enteros.pop(0) # Asignarle a la variable entero el primer elemento de la lista en cada ciclo
    
    if entero % 2 == 0:
        print(f"{entero} es par")
````
