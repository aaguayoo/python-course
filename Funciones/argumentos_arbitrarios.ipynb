# Argumentos arbitrarios

## Posicionales

Supongamos que necesitamos crear una función que regrese la suma de todos los ingresos registrados para una persona. Sabemos de antemano que la persona cuenta actualmente con dos ingresos, por lo tanto podríamos crear una función como la siguiente:

````python
def calcula_ingreso_total(ingreso_1: float, ingreso_2: float) -> float:
    return ingreso_1 + ingreso_2
````

````python
calcula_ingreso_total([2,3,3], [3,3])
````

Tiempo después, sabemos que la persona ahora cuenta con un nuevo ingreso, por lo que tenemos que modificar ahora nuestra función para que admita un nuevo argumento:

````python
def calcula_ingreso_total(
    ingreso_1: float, 
    ingreso_2: float, 
    ingreso_3: float,
) -> float:
    return ingreso_1 + ingreso_2 + ingreso_3
````

Una semana después, la persona ha dejado de recibir el primer ingreso, lo que supondría que necesitamos modificar neuvamente nuestra función.

````python
def calcula_ingreso_total(
    ingreso_2: float, 
    ingreso_3: float,
) -> float:
    return ingreso_2 + ingreso_3
````

Continuar por ese camino resulta inviable, y más si lo que se busca es una aplicación o código que el usuario pueda manejar sin necesidad de estar cambiando constantemente. Por esot, necesitamos encontrar una forma en que podamos adminitr tantos argumentos como sean necesarios. 

Una primera solución es solicitar un único argumento que contenga todos los ingreso, ya sea un conjunto o, mucho mejor, una lista o diccionario.

````python
def calcula_ingreso_total(
    lista_ingresos: list[float],
) -> float:
    return sum(lista_ingresos)
````

El problema con esta solución es que agrega un paso al momento de enviar los datos, el cuál sería almacenar los ingresos en una lista antes de ser pasados a nuestra función

````python
ingresos = [10_000, 4_000, 15_000] # Paso extra que usa memoria
calcula_ingreso_total(ingresos)
````

Y en realidad, esto no resuelve el problema de poder suministrar una cantidad arbitraria de argumentos a nuestra función, ya que vamos a tener que estar llenando la lista constantemente y nuestra función sólo recibirá un argumento cuando el problema era saber si se podían mandar N argumentos arbitrarios.

Para resolver el problema se puede utilizar la siguiente notación:

````python
def calcula_ingreso_total(
    *args
) -> float:
    # Se puede agregar sentencia para validar args
    return sum(args)
````

**NOTA**: El nombre _args_ se utiliza por convención, pero en principio puede llamarse como sea, siempre que se le incluya el `*` al inicio.

La notación `*args` nos permitirá recibir tantos argumentos posicionales como sean pasados a la función y todos serán almacenados de forma dinámica dentro de la lista con nombre `args`. 

````python
calcula_ingreso_total(1)
````

````python
calcula_ingreso_total(2,3,4)
````

````python
calcula_ingreso_total(4,5,4,3,2,3,4,5,2,3,4,2,3)
````

**NOTA**: Los argumenos que se pasen a `*args` deben tener cierta consistencia predefinida por el usuario (e.g., si se van a sumar, pues que todos sean int o float)

````python
def calcula_ingreso_total(
    *args
) -> float:
    # Se puede agregar sentencia para validar args
    print(args)
````

````python
calcula_ingreso_total(1,2,3,4,5,56,6,5)
````

## Pre-definidos

De forma análoga, es posible suministrar tantos argumentos pre-definidos a la función a traves'de la siguiente notación:

````python
# Función sin kwargs
def color_rgb(rojo: tuple = (255,0,0), azul: tuple = (0, 0, 255)) -> dict:
    print({"rojo": rojo, "azul": azul})
````

````python
color_rgb()
````

````python
def colores_rgb(**kwargs) -> dict:
    print(kwargs)
````

````python
colores_rgb(green=(0, 255, 0))
````

````python
colores_rgb(yellow=(255,255,0), green=(0,255,0))
````

````python
colores_rgb(
    yellow=(255,255,0), 
    green=(0,255,0),
    red=(255,0,0),
)
````

## Funciones de argumentos dinámicos

A través de los argumenos args y kwargs es posible definir funciones que admitan cualquier cantidad de argumentos de cualquier tipo, tanto posicionales como pre-definidos:

```python
def funcion_arbitrario(*args, **kwargs):
    ...
```

Funciones que no tienes ni idea de qué van a recibir, pero que si reciben algo sabes que debes de hacer algo con sus valores pre-definidos (kwargs) y con sus posicionales (args).

### Función de argumentos dinámicos con decorador

````python
# Creamos el decorador
def imprimir_saludo(funcion: callable):
    def wrapper(*args, **kwargs):
        print("Hola, estas a punto de correr una función en Python. Éxito")
        resultado = funcion(*args, **kwargs)
        print(resultado)
        print("Tu función corrió exitosamente")
        return resultado
    
    return wrapper
````

````python
@imprimir_saludo
def calcula_ingresos(*args):
    return sum(args)

@imprimir_saludo
def regresa_cadena_texto(*args):
    return "".join(args)
````

````python
calcula_ingresos(1,2,3,4,5,6,2,3)
````

````python
regresa_cadena_texto("a","b","c","d")
````
