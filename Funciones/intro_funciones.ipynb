# Introducción a funciones

Es común que cuando comienza a crecer un código ,existan sentencias declaradas que comiencen a repetirse, más aún, que sean bloques de sentencias y no una única sentencia. Cuando estamos en una situación como la mencionada, entonces es necesario considerar la creación de _funciones_.

Entenderemos por _función_ a un bloque de sentencias previamente declarado y nombrado para ser utilizado en el momento en que se requiera, a diferencia de las estructuras de flujo que son declaradas en el código en un momento determinado para su uso.

Cuando una función es declarada, Python asignará un nombre de referencia a dicho código, sin embargo, si se espera tener un resultado utilizando la función, esta tendrá que ser invocada correctamente para su ejecución.

La declaración de funciones tiene la siguiente estructura y, es importante recordar, que de preferencia, no se escriban dentro del código, sino en algún archivo o espacio aparte:

```python
def nombre_de_la_funcion(argmuentos):
    sentencias_asociadas
```

**Notas**: 
1. El nombre de la función siempre debe ser lo más explícito posible respecto de lo que hace la función y, de preferencia, debe iniciar con un verbo.
2. Todas las variables declaradas dentro de una función se consideran variables locales. Es decir, no pueden ser accedidas desde ninguna otra parte del código, y desde ninguna otra función.

Supongamos que se quiere obtener el cuadarado de la suma de dos números, entonces, podemos declarar:

````python
def calcular_cuadrado_de_suma(numero_1, numero_2):
    suma = numero_1 + numero_2
    return suma**2
````

Analicemos la función _calcular_cuadrado_de_suma_. La partícula **def** es siempre obligatoria para la definición de funciones. El nombre es la referencia con la cuál podremos invocar la función, y `numero_1` y `numero_2` son los argumentos (valores de entrada de la función). Después, de forma indentada, las declaraciones asociadas a la función. Nótese que se ha utilizado la partícula **return**, de momento no entraremos en detalles sobre su utilidad.

## Funciones simples

Cuando desarrollmos funciones, nuestra forma de irlas asimilando en todos sus aspectos puede ser de forma gradual, es posible simplificar su declaración de moa a que sean comprensibles sus componentes.

Simplifiquemos la declaración reduciendo el proceso a declarla y ejecutar una sencilla sentencia.

````python
# Función que no recibe argumentos y que no usa return
def imprime_hola_mundo():
    print("Hola, mundo!")
````

**NOTA**: Declarar una función no implica que será ejecutada, es necesario que sea invocada para ejecutar las sentencias declaradas dentro de la función.

````python
# Invocamos la función
imprime_hola_mundo()
````

````python
# ¿Qué pasa si no se ponen los paréntesis?
imprime_hola_mundo
````

````python
type(imprime_hola_mundo())
````

## Return

Por el momento hemos conseguido que nuestra función devuelva un valor en pantalla, sin embargo, no quiere decir que lo mostrado a través de la función _print_ sea información que podamos utilizar.

````python
saludo = print("Hola mundo") # Aunque se imprima el Hola mundo, esta salida no se va a guardar en saludo
````

````python
saludo
````

````python
saludo = imprime_hola_mundo() # La función no regresa nada util, que se pueda guardar en una variable.
````

````python
print(saludo)
````

Por lo que esperamos de algún modo recuperar el mensaje dentro de nuestra fución y poder almacenarlo en alguna variable, para eso existe la partícula **return**, la cual nos permite devolver un valor que puede ser colectad na vez que se invoca la función. Volviento a nuestra función `imprime_hola_mundo`, queremos regresar el mensaje:

````python
def regresa_hola_mundo():
    return "Hola, mundo!"
````

````python
# Invoncamos de nuevo
saludo = regresa_hola_mundo()
print(saludo)
````

````python
imprime_hola_mundo() # Esta invocación ya se puede ver como una variable
````

````python
# Se puede usar para definir alguna otra variable
saludo = imprime_hola_mundo()
````

## Llamar función dentro de otra función

Crear una función completa, que reciba dos funciones previas ya utilizadas

````python
def calcula_promedio(numero_1, numero_2):
    sumatory = numero_1 + numero_2
    den = 2
    promedio = sumatory / den
    return promedio
````

````python
def realiza_calculo_extra(numero_1, numero_2, promedio):
    # Restamos numero_1 a promedio
    resta = promedio - numero_1

    # Multiplicamos resultado por promedio entre numero_2
    resultado_final = resta * promedio / numero_2
    
    return resultado_final
````

````python
def realiza_calculo_completo_y_promedio(numero_1, numero_2):
    promedio = calcula_promedio(numero_1, numero_2)
    resultado = realiza_calculo_extra(numero_1, numero_2, promedio)
    return resultado, promedio
````

````python
numero_1 = 27
numero_2 = 35

resultado, promedio = realiza_calculo_completo_y_promedio(numero_1, numero_2)

print(resultado)
print(promedio)
````
