# Tipos d argumenos para funciones en Python

El declarar funciones adquiere una condición más dinámica cuando consideramos el caso donde nuestra función depende de uno o varios valores de entrada que pueden ser procesados para obtener un resultado.

Supongamos una función la cuál recibe como entrada un número y regresa su valor elevado al cuadarado, necesitaríamos indicar a nustra función que requiere un parámetro de ntrada el cuál puede ser declarado de la sguiente manera:

````python
def elevar_al_cuadrado(numero):
    return numero**2
````

Nótese que hemos agregado `numero` entre los parénteis de la declaración de nuestra función, a estos elementos agregados se les conoce como _argumentos_. Los argumentos son los parámetros de entrada que son admitiso por la función eal momento de ser invocada. Los argumentos serán variables locales que tendrán alcance únicamente dentro de la función. Si se requiere acceder (o regresar) alguno de las variables locales de lafunción, siempre se puede usar la párticula _return_.

## Tipos de argumentos

En cuanto decidimos hacer que neustra función reciba valors de entrada, necesitamos determinar qué tipo de argumnos va a recibir. En **Python** se puden emplear dos tipos de argumentos: _posicionales_ y los _pre-determinados_.

### Argumentos posicionales

Ls argumentos posicionales son aquellos que durante la invocación de la función, los valores son asignados a las variables locales según el orden en que son suministrados.

Supongamos una función que admite dos argumentos:
```python
def suma_numeros(numero_1, numero_2):
    return numero_1 + numero_2
```

Al momento de invocar la función suminstraremos los valores para cada argumento, tal cómo fueron definidos en la declaración de la función

```python
suma_numeros(1,2)
```

en este caso, según la posición de cada argumento tendriamos que la posición 0 `numero_1 = 1` y en la posición 1 
`numero_2 = 2`.

Otra característica importante de los argumentos posicionales es que son obligatorios, es decir que se deben suministrar tantos valores en la invocación como argumentos posicionales se requieran, de lo contrario se levantará un error.

````python
def suma_numeros(numero_1, numero_2):
    return numero_1 + numero_2
````

````python
# Pasamos 1 argumento cuando la función tiene 2 posicionales
suma_numeros(10)
````

````python
# Pasamos mas argumenots de los que recibe
suma_numeros(1,2,3)
````

También se puede evitar la ambigüedad de la posición escribiendo en la invocación justamente a que argumento posicional va cada uno de los datos de entrada

````python
suma_numeros(
    numero_1=123,
    numero_2=2,
)
````

### Argumentos pre-definidos

Por otro lado, existe una forma de declarar argumentos que no requieren que se les sea asignado un valor de forma obligatoria, sino que puedan mantener un valor por defecto.

```python
def suma_numeros(numero_1, numero_2, numero_3=0):
    return numero_1 + numero_2 + numero_3
```

Nótese que durante la declaración de argumentos, al argumento `numero_3` se le ha asignado un valor por defecto = 0. 

**NOTA**: Los argumentos predefiniso nunca pueden ir antes de los argumentos posicionales.

````python
# Esto está mal y da un error de sintaxis
def suma_numeros(numero_2, numero_3, numero_1=0, numero_4=5, numero_6=8, numero_5=10):
    return numero_1 + numero_2 + numero_3 + numero_4 + numero_5
````

- str
- int
- float
- list
- dict
- tuple
- set

- np.ndarray (NumPy)
- torch.Tensor (Torch)
- Image (PIL)
