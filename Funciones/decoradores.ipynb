# Decoradores

Antes de hablar de decoradores, es necesario mencionar el concepto de _funciones internas_, el cuál hace referncia al hecho de poder declarar funciones dentro de otras funciones. La pregunta natural sería ¿por qué quisiera declarar una función dentro de otra?. Existen diversas razones, pero quizás la principal es que las funciones internas únicamente serán empleadas dentro de la rutina principal de la función contenedora.

De hecho, cuando se comienza en las prácticas de **factorización** (factor, refactor) es común encontrar funciones dentro de funciones, ya que ayuda a simplificar el entendimiento de una función extensa. 

Ejemplo: Supongamos que se quiere realizar una función que pueda calcular el impuesto sobre un total, dependiento del tipo de impuesto solicitado es el tipo de cálculo a realizar:

```python
def calcular_impuesto(total: float, impuesto: str) -> float:
    if impuesto == "iva":
        return total * 0.16
    elif impuesto == "isr":
        return total * 0.18
    else:
        return total
```

Aunque los procesos del cálculo de impuesto son sencillos, pueden ser mejor manejados si creamos funciones donde puedan ser realizados por separdo:

```python
def calcular_impuesto(total: float, impuesto: str) -> float:
    
    def calcula_iva(total: float):
        return total * 0.16
   
    def calcula_isr(total: float):
        return total * 0.18
    
    if impuesto == "iva":
        return calcula_iva(total)
    elif impuesto == "isr":
        return calcula_isr(total)
    else:
        return total
```

```python
calcular_impuesto(100, impuesto="iva")
```

Aunque el desarrollo se vuelve más lento y el número de líneas se extiende, resulta más sencillo dar mantenimiento a un código factorizado ya que se garantiza la escalabilidad
